# Análisis Exploratorio de Datos - Dataset PISA
## Estudio de un país específico

Este notebook realiza un análisis exploratorio completo del dataset PISA para identificar:
- Cantidad de variables y registros
- Tipos de datos y sus cantidades
- Valores faltantes por variable
- Estadísticas descriptivas
- Distribuciones y análisis adicionales

```python
# Importar librerías necesarias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
import warnings
warnings.filterwarnings('ignore')

# Configuración de visualización
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## 1. Carga de Datos

```python
# Cargar el dataset PISA
# Ajusta la ruta según tu archivo
ruta_archivo = './../data_collection_peru_csv/STU_QQQ_PERU.csv'  # Puede ser .csv, .xlsx, .sav, etc.

# Descomenta la línea apropiada según tu formato de archivo:
df = pd.read_csv(ruta_archivo)
# df = pd.read_excel(ruta_archivo)
# df = pd.read_spss(ruta_archivo)  # Para archivos .sav

print(f"✓ Dataset cargado exitosamente")
print(f"Primeras filas del dataset:")
display(df.head())
```

## 2. Información General del Dataset

```python
# Cantidad de registros y variables
n_registros, n_variables = df.shape

print("="*60)
print("RESUMEN GENERAL DEL DATASET")
print("="*60)
print(f"\n📊 Cantidad de registros (filas): {n_registros:,}")
print(f"📊 Cantidad de variables (columnas): {n_variables:,}")
print(f"\n💾 Tamaño en memoria: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
print("="*60)
```

## 3. Tipos de Datos y sus Cantidades

```python
# Contar tipos de datos
tipos_datos = df.dtypes.value_counts()

print("\n" + "="*60)
print("TIPOS DE DATOS EN EL DATASET")
print("="*60)
for tipo, cantidad in tipos_datos.items():
    print(f"  {str(tipo):15} : {cantidad:3} variables ({cantidad/n_variables*100:.1f}%)")

# Visualización de tipos de datos
plt.figure(figsize=(10, 6))
tipos_datos.plot(kind='bar', color='steelblue')
plt.title('Distribución de Tipos de Datos', fontsize=14, fontweight='bold')
plt.xlabel('Tipo de Dato')
plt.ylabel('Cantidad de Variables')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Clasificar variables por tipo
variables_numericas = df.select_dtypes(include=[np.number]).columns.tolist()
variables_categoricas = df.select_dtypes(include=['object', 'category']).columns.tolist()
variables_datetime = df.select_dtypes(include=['datetime64']).columns.tolist()

print(f"\n📈 Variables numéricas: {len(variables_numericas)}")
print(f"📝 Variables categóricas: {len(variables_categoricas)}")
print(f"📅 Variables de fecha/hora: {len(variables_datetime)}")
```

## 4. Análisis de Valores Faltantes

```python
# Calcular valores faltantes
valores_faltantes = pd.DataFrame({
    'Variable': df.columns,
    'Faltantes': df.isnull().sum(),
    'Porcentaje': (df.isnull().sum() / len(df)) * 100,
    'Tipo': df.dtypes
}).sort_values('Faltantes', ascending=False)

# Filtrar solo variables con valores faltantes
valores_faltantes_con_missing = valores_faltantes[valores_faltantes['Faltantes'] > 0]

print("\n" + "="*60)
print("ANÁLISIS DE VALORES FALTANTES")
print("="*60)
print(f"\nVariables con valores faltantes: {len(valores_faltantes_con_missing)} de {n_variables}")
print(f"Variables completas (sin faltantes): {n_variables - len(valores_faltantes_con_missing)}")

if len(valores_faltantes_con_missing) > 0:
    print(f"\n📊 TOP 20 variables con más valores faltantes:\n")
    display(valores_faltantes_con_missing.head(20).style.format({
        'Faltantes': '{:,.0f}',
        'Porcentaje': '{:.2f}%'
    }).background_gradient(subset=['Porcentaje'], cmap='Reds'))
    
    # Visualización de valores faltantes (top 20)
    top_missing = valores_faltantes_con_missing.head(20)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    
    # Gráfico de barras
    ax1.barh(range(len(top_missing)), top_missing['Porcentaje'], color='coral')
    ax1.set_yticks(range(len(top_missing)))
    ax1.set_yticklabels(top_missing['Variable'], fontsize=8)
    ax1.set_xlabel('Porcentaje de Valores Faltantes (%)')
    ax1.set_title('Top 20 Variables con Más Valores Faltantes', fontweight='bold')
    ax1.invert_yaxis()
    
    # Gráfico de distribución
    ax2.hist(valores_faltantes_con_missing['Porcentaje'], bins=30, color='steelblue', edgecolor='black')
    ax2.set_xlabel('Porcentaje de Valores Faltantes (%)')
    ax2.set_ylabel('Cantidad de Variables')
    ax2.set_title('Distribución de Valores Faltantes', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
else:
    print("\n✓ No hay valores faltantes en el dataset")
```

```python
# Categorizar variables según nivel de missing
categorias_missing = pd.cut(valores_faltantes['Porcentaje'], 
                            bins=[-0.1, 0, 5, 20, 50, 100],
                            labels=['Sin faltantes', 'Bajo (0-5%)', 'Medio (5-20%)', 
                                   'Alto (20-50%)', 'Muy Alto (>50%)'])

resumen_missing = categorias_missing.value_counts().sort_index()

print("\n📊 Resumen por nivel de valores faltantes:\n")
for categoria, cantidad in resumen_missing.items():
    print(f"  {str(categoria):20} : {cantidad:3} variables ({cantidad/n_variables*100:.1f}%)")
```

## 5. Estadísticas Descriptivas - Variables Numéricas

```python
if len(variables_numericas) > 0:
    print("\n" + "="*60)
    print("ESTADÍSTICAS DESCRIPTIVAS - VARIABLES NUMÉRICAS")
    print("="*60)
    
    # Estadísticas básicas
    estadisticas = df[variables_numericas].describe().T
    estadisticas['missing'] = df[variables_numericas].isnull().sum()
    estadisticas['missing_pct'] = (df[variables_numericas].isnull().sum() / len(df)) * 100
    
    print(f"\nEstadísticas de las primeras 10 variables numéricas:\n")
    display(estadisticas.head(10).style.format({
        'mean': '{:.2f}',
        'std': '{:.2f}',
        'min': '{:.2f}',
        '25%': '{:.2f}',
        '50%': '{:.2f}',
        '75%': '{:.2f}',
        'max': '{:.2f}',
        'missing_pct': '{:.2f}%'
    }))
else:
    print("No hay variables numéricas en el dataset")
```

## 6. Análisis de Variables Categóricas

```python
if len(variables_categoricas) > 0:
    print("\n" + "="*60)
    print("ANÁLISIS DE VARIABLES CATEGÓRICAS")
    print("="*60)
    
    # Análisis de cardinalidad
    cardinalidad = pd.DataFrame({
        'Variable': variables_categoricas,
        'Categorías_Únicas': [df[col].nunique() for col in variables_categoricas],
        'Valores_Faltantes': [df[col].isnull().sum() for col in variables_categoricas],
        'Pct_Faltantes': [(df[col].isnull().sum() / len(df)) * 100 for col in variables_categoricas]
    }).sort_values('Categorías_Únicas', ascending=False)
    
    print(f"\nPrimeras 15 variables categóricas por cantidad de categorías únicas:\n")
    display(cardinalidad.head(15).style.format({
        'Pct_Faltantes': '{:.2f}%'
    }))
    
    # Mostrar frecuencias de algunas variables categóricas de interés
    # (Ajusta estas según las variables de tu dataset PISA)
    variables_interes = [col for col in variables_categoricas[:5] if df[col].nunique() < 50]
    
    if variables_interes:
        print(f"\n📊 Distribución de frecuencias (primeras {len(variables_interes)} variables):")
        for col in variables_interes:
            print(f"\n{col}:")
            freq = df[col].value_counts().head(10)
            for valor, cantidad in freq.items():
                pct = (cantidad / len(df)) * 100
                print(f"  {str(valor)[:40]:40} : {cantidad:6,} ({pct:5.2f}%)")
else:
    print("No hay variables categóricas en el dataset")
```

## 7. Matriz de Correlación (Variables Numéricas)

```python
if len(variables_numericas) > 0:
    # Calcular correlación solo con variables que tengan suficientes datos
    vars_para_corr = [col for col in variables_numericas 
                      if df[col].isnull().sum() / len(df) < 0.5]
    
    if len(vars_para_corr) >= 2:
        # Limitar a las primeras 30 variables para visualización
        vars_para_corr = vars_para_corr[:30]
        
        correlacion = df[vars_para_corr].corr()
        
        plt.figure(figsize=(14, 12))
        sns.heatmap(correlacion, annot=False, cmap='coolwarm', center=0,
                   square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
        plt.title('Matriz de Correlación (primeras 30 variables numéricas)', 
                 fontsize=14, fontweight='bold', pad=20)
        plt.tight_layout()
        plt.show()
        
        # Encontrar correlaciones altas
        print("\n📊 Pares de variables con correlación alta (|r| > 0.7):\n")
        high_corr = []
        for i in range(len(correlacion.columns)):
            for j in range(i+1, len(correlacion.columns)):
                if abs(correlacion.iloc[i, j]) > 0.7:
                    high_corr.append({
                        'Variable_1': correlacion.columns[i],
                        'Variable_2': correlacion.columns[j],
                        'Correlación': correlacion.iloc[i, j]
                    })
        
        if high_corr:
            df_high_corr = pd.DataFrame(high_corr).sort_values('Correlación', 
                                                               key=abs, 
                                                               ascending=False)
            display(df_high_corr.head(10))
        else:
            print("No se encontraron correlaciones altas entre las variables analizadas")
```

## 8. Identificación de Valores Atípicos (Outliers)

```python
if len(variables_numericas) > 0:
    print("\n" + "="*60)
    print("ANÁLISIS DE VALORES ATÍPICOS")
    print("="*60)
    
    # Análisis usando método IQR
    outliers_info = []
    
    for col in variables_numericas[:20]:  # Primeras 20 variables
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
        
        if len(outliers) > 0:
            outliers_info.append({
                'Variable': col,
                'Num_Outliers': len(outliers),
                'Pct_Outliers': (len(outliers) / len(df)) * 100,
                'Min_Outlier': outliers.min(),
                'Max_Outlier': outliers.max()
            })
    
    if outliers_info:
        df_outliers = pd.DataFrame(outliers_info).sort_values('Num_Outliers', ascending=False)
        print(f"\nVariables con outliers detectados (método IQR):\n")
        display(df_outliers.head(15).style.format({
            'Pct_Outliers': '{:.2f}%',
            'Min_Outlier': '{:.2f}',
            'Max_Outlier': '{:.2f}'
        }))
    else:
        print("\nNo se detectaron outliers significativos en las variables analizadas")
```

## 9. Resumen Ejecutivo del Dataset

```python
print("\n" + "="*70)
print("RESUMEN EJECUTIVO DEL ANÁLISIS")
print("="*70)

print(f"\n📊 DIMENSIONES:")
print(f"   • Total de registros: {n_registros:,}")
print(f"   • Total de variables: {n_variables:,}")

print(f"\n📈 TIPOS DE DATOS:")
print(f"   • Variables numéricas: {len(variables_numericas):,}")
print(f"   • Variables categóricas: {len(variables_categoricas):,}")
print(f"   • Variables de fecha/hora: {len(variables_datetime):,}")

print(f"\n❌ VALORES FALTANTES:")
total_missing = df.isnull().sum().sum()
pct_total_missing = (total_missing / (n_registros * n_variables)) * 100
print(f"   • Total de valores faltantes: {total_missing:,} ({pct_total_missing:.2f}% del dataset)")
print(f"   • Variables con valores faltantes: {len(valores_faltantes_con_missing):,}")
print(f"   • Variables sin valores faltantes: {n_variables - len(valores_faltantes_con_missing):,}")

if len(valores_faltantes_con_missing) > 0:
    var_max_missing = valores_faltantes_con_missing.iloc[0]
    print(f"   • Variable con más faltantes: {var_max_missing['Variable']} ({var_max_missing['Porcentaje']:.2f}%)")

print(f"\n💾 MEMORIA:")
print(f"   • Tamaño total en memoria: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

print("\n" + "="*70)
print("✓ Análisis completado exitosamente")
print("="*70)
```
