```python
# Exploración general
def basic_inspection(name_df: str, df):
    print(name_df + " - Tamaño del dataset:", df.shape)
    print("\nTipos de datos:")
    print(df.dtypes.value_counts())

# Columnas que estan en ambos DataFrames
def both_dataframes_columns(main_df, new_df, name_ignore_column: str = ""):
    if name_ignore_column == "":
        common_columns = set(main_df.columns).intersection(set(new_df.columns))
    else:
        common_columns = set(main_df.columns).intersection(set(new_df.columns)) - {name_ignore_column}

    return common_columns

def merge_df(main_df, new_df, common_columns = [], key_value = []):
    # Elimina esas columnas del segundo dataset
    new_df_sin_duplicadas = new_df.drop(columns=common_columns)

    # Merge usando el identificador del escuela
    return main_df.merge(new_df_sin_duplicadas, on=key_value, how="left")

def veridicar_var_similar(main_df, new_df, common_columns, colum_relation: str):
    for col in common_columns:
        merged_check = main_df[[colum_relation, col]].merge(
            new_df[[colum_relation, col]],
            on=colum_relation,
            how='inner',
            suffixes=('_stu', '_sch')
        )

        # Comparación que trata NaN como iguales
        different = merged_check[
            ~merged_check[f"{col}_stu"].fillna('MISSING').eq(
            merged_check[f"{col}_sch"].fillna('MISSING'))
        ]

        print(f"Columna: {col}")
        if different.empty:
            print("✅ Los valores son iguales en todas las filas coincidentes.\n")
        else:
            print(f"❌ Hay diferencias en {len(different)} filas.\n")

def show_diferent_values(main_df, new_df, colum_relation: str, colum_verificate: str, _main_df: str = '_stu', _new_df: str = '_sch'):
    col = colum_verificate

    test_merged_check = main_df[[colum_relation, col]].merge(
        new_df[[colum_relation, col]],
        on=colum_relation,
        how='inner',
        suffixes=(_main_df, _new_df)
    )

    different = test_merged_check[test_merged_check[f"{col}" + _main_df] != test_merged_check[f"{col}" + _new_df]]

    # Mostrar los diferentes
    print(different)
```

```python
basic_inspection("STU_SCH_df", STU_SCH_df)
basic_inspection("CRT_COG_df", CRT_COG_df)
```

```python
common_columns = both_dataframes_columns(STU_SCH_df, CRT_COG_df, "CNTSTUID")
print(common_columns)
```

```python
veridicar_var_similar(STU_SCH_df, CRT_COG_df, common_columns, "CNTSTUID")
```

```python
STU_SCH_CRT_df = merge_df(STU_SCH_df, CRT_COG_df, common_columns, ["CNTSTUID"])
basic_inspection("STU_SCH_CRT_df", STU_SCH_CRT_df)
```

```python
# Coleccion de datos estudiante
STU_ALL_df = pd.read_csv("../data_collection_peru_csv/STU_ALL_PERU.csv")

# Colleccion de datos - Contexto escolar
SCH_QQQ_df = pd.read_csv("../data_collection_peru_csv/SCH_QQQ_PERU.csv")
```

```python
# Exploración general STU_ALL
print("STU_ALL - Tamaño del dataset:", STU_ALL_df.shape)
print("\nTipos de datos:")
print(STU_ALL_df.dtypes.value_counts())

# Exploración general SCH_QQQ
print("SCH_QQQ - Tamaño del dataset:", SCH_QQQ_df.shape)
print("\nTipos de datos:")
print(SCH_QQQ_df.dtypes.value_counts())
```

```python
# Columnas que estan en ambos DataFrames (STU_ALL_df + SCH_QQQ_df)
common_columns = set(STU_ALL_df.columns).intersection(set(SCH_QQQ_df.columns)) - {"CNTSCHID"}
print(common_columns)
```

```python
# MERGE POR PARTES
# Elimina esas columnas del segundo dataset
SCH_QQQ_df_sin_duplicadas = SCH_QQQ_df.drop(columns={'STRATUM', 'ADMINMODE', 'NatCen', 'REGION', 'SUBNATIO', 'CYC', 'CNTRYID', 'CNT', 'OECD'})

# Merge usando el identificador del escuela
# STU_ALL = STU_COG_df_STU_QQQ_df.merge(SCH_QQQ_df_sin_duplicadas, on=["CNTSTUID"], how="left")
STU_SCH_ALL = pd.merge(
    STU_ALL_df,      # el dataset de estudiantes
    SCH_QQQ_df_sin_duplicadas,     # dataset de escuelas
    on='CNTSCHID',  # clave única
    how='inner',
    suffixes=('_stu', '_sch')  # esto mantiene ambas columnas si hay nombres duplicados
)
```

```python
STU_SCH_ALL.to_csv("../data_collection_peru_csv/STU_SCH_ALL_PERU.csv", index=False)

print("¡Listo! Se generaron los archivos Excel.")
```
